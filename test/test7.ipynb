练习题 4：基于纯循环的朴素卷积实现（编程题）

题目：

《深度学习入门》中使用了高阶技巧 im2col 来加速卷积运算。但为了彻底理解卷积滑动的底层逻辑，请不使用任何现成的框架，仅用 Python 原生的 for 循环和基本的 NumPy 操作，实现一个“单通道输入、单滤波器、无填充、步幅为 1”的朴素卷积前向传播函数。

```python
import numpy as np

def naive_conv2d(x, w):
    """
    x: 输入图像矩阵, shape (H, W)
    w: 滤波器权重矩阵, shape (FH, FW)
    """
    H, W = x.shape
    FH, FW = w.shape

    # 1. 计算输出特征图的形状
    OH = H - FH + 1
    OW = W - FW + 1

    # 初始化输出矩阵
    out = np.zeros((OH, OW))

    # 2. 双重循环模拟滑动窗口
    for i in range(OH):
        for j in range(OW):
            # 切片提取当前窗口的数据
            x_window = x[i : i+FH, j : j+FW]
            # 元素对应相乘并求和
            out[i, j] = np.sum(x_window * w)

    return out

# 测试代码 (使用练习题 2 的数据进行验证)
X = np.array([[1, 2, 1],
              [0, 1, 2],
              [3, 0, 1]])
W = np.array([[2, 0],
              [1, -1]])

result = naive_conv2d(X, W)
print("代码运行结果:\n", result)
# 预期输出与练习题 2 手算结果一致：
# [[1. 3.]
#  [3. 1.]]
```
